```python
df.info()
```

```python
# Respostas:

# Existem valores nulos? Não
# Quais colunas são numéricas e quais são texto? Colunas numéricas: 5, 7, 8, 9, 10, 12; o restante das colunas está em formato "object"
# Existe coluna de data? Sim, colunas 0 e 1
```

```python
df["data_inicial"] = pd.to_datetime(df["data_inicial"])
df["data_final"] = pd.to_datetime(df["data_final"])

df.info()
```

```python
#df["PREÇO MÉDIO REVENDA"] = df["PREÇO MÉDIO REVENDA"].str.replace(",", ".").astype(float)

# o código acima não é necessário pois os valores já estão com ponto ao invés de vírgula
```

```python
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")

df.head(5)
```

# Etapa 3 - Exploração inicial (EDA básico)

```python
df["produto"] = df["produto"].str.replace("Ó", "O")
```

```python
# Produtos pesquisados

df["produto"].unique()
```

```python
# Quantidade de produtos pesquisados

df["produto"].nunique()
```

```python
# Estados pesquisados

df["estado"].unique()
```

```python
# Quantidade de estados pesquisados

df["estado"].nunique()
```

```python
# Número de registros em cada estado

df["estado"].value_counts()
```

```python
# Preco médio geral

pmr = df.groupby("produto")["preço_médio_revenda"].mean()

print(pmr)
```

# Etapa 4 - Transformação (Engenharia simples)

```python
# Criando coluna "ANO" a partir da data inical

df["ANO"] = df["data_inicial"].dt.year

df.head(5)
```

```python
# Agrupar preco médio por ano

preco_produto_ano = df.groupby("ANO")["preço_médio_revenda"].mean()

print(preco_produto_ano)
```

```python
# Agrupar por combustivel

preco_produto_comb = df.groupby("produto")["preço_médio_revenda"].mean()

print(preco_produto_comb)
```

# Insight

```python
gnv = df.groupby("produto")["unidade_de_medida"].first()

print(gnv)
```

### O GLP acaba inflando o preço médio dos produtos se analisarmos toda a série pois seu preço por unidade de medida acaba sendo, em ordem de grandeza, maior do que os outros. A unidade de medidda é reais por m3, isso afeta a análise de preço, elevando a média - pode acontecer no caso do GVN também. O correto é padronizarmos todas as unidades de medida.

# Etapa 5 - Análise um pouco aprofundada


```python
# Qual combustível ficou mais caro?

df.groupby(["ANO", "produto"])["preço_médio_revenda"].mean()
```

```python
# Qual estado tem combustível mais caro em média?

df.groupby("estado")["preço_médio_revenda"].mean().sort_values(ascending=False)
```

```python
# Evolução de um combustível específico (gasolina comum)

gasolina = df[df["produto"] == "GASOLINA COMUM"]

gasolina.groupby("ANO")["preço_médio_revenda"].mean()
```

# Validação final

```python
# Evolução temporal por produto

evolucao = df.groupby(["ANO", "produto"])["preço_médio_revenda"].mean()

evolucao.loc[[2004, 2010, 2015, 2020, 2021]]
```

```python
# Tendência isolada (gasolina comum)

gasolina = df[df["produto"] == "GASOLINA COMUM"]

gasolina.groupby("ANO")["preço_médio_revenda"].mean()
```

```python
# Top 5 estados mais caros

df_sem_glp = df[df["produto"] != "GLP"]

df_sem_glp.groupby("estado")["preço_médio_revenda"].mean().sort_values(ascending = False).head()
```
